```python
matrix = []

print("Enter 9 elements:")

# Input 3x3 matrix
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix.append(row)

# Get row and column index
row = int(input("Enter row index (0-2): "))
col = int(input("Enter column index (0-2): "))

# Access element
print("Element =", matrix[row][col])
```
